```python
import pandas as pd

data = pd.read_csv("/content/kmanpub.csv")
```

```python
# I does this contain the target keywords:
keywords = ["Machine learning", "Psychology"]

labels = []

for i in range(len(data)):

  buffer = []

  for keyword in keywords:
    if keyword in [keyword.strip() for keyword in str(data["keywords"][i]).split(";")]:
      buffer.append(1)
    else:
      buffer.append(0)

  if 0 in buffer:
    labels.append(0)
  else:
    labels.append(1)


print(labels)
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
from langchain_core.documents import Document

# Convert the 'keywords' column into a list of Document objects, including the original index as metadata
documents = [Document(page_content=str(keyword), metadata={'row_id': idx}) for idx, keyword in data['keywords'].items()]

db = FAISS.from_documents(documents,
                          HuggingFaceEmbeddings(model_name='BAAI/bge-base-en-v1.5'))
```

```python
retriever = db.as_retriever(
    search_type="similarity",
    search_kwargs={'k': 4}
)
```

```python
def RAG(Question):
  retrieved_docs_direct = db.similarity_search(Question, k=4)

  context = "<Articles found by the search engine>\n"

  for doc in retrieved_docs_direct:
    row = doc.metadata.get('row_id')

    context += "Title:\n"
    context += str(data["title"][row]) + "\n"
    context += "Authors:\n"
    context += str(data["authors"][row]) + "\n"
    context += "Publication date:\n"
    context += str(data["date_published"][row]) + "\n"

  context += "</Articles found by the search engine>"

  return context
```

```python
from langchain_core.prompts import PromptTemplate

question = "Can you look me up articles about the use of machine learning in psychology?"

context = RAG(question)

print(context)
```

```python
from scipy.spatial.distance import cosine
embedder = HuggingFaceEmbeddings(model_name='BAAI/bge-base-en-v1.5')
def RAG_eval(keywords: list, embedder):
  question = f"Hey, can you give me some reading about {keywords[0]} and {keywords[1]}"

  keywords_string = "; ".join(keywords)
  keywords_embedding = embedder.embed_query(keywords_string)

  retrieved_docs_direct = db.similarity_search(question, k=4)

  distances = []

  for doc in retrieved_docs_direct:

    row = doc.metadata.get('row_id')
    retrieved_keywords = data["keywords"][row]
    retrieved_keywords_embedding = embedder.embed_query(retrieved_keywords)
    distance = cosine(retrieved_keywords_embedding, keywords_embedding)
    distances.append(distance)

  return distances
```

```python
RAG_eval(["Psychology", "Machine learning"], embedder)
```
